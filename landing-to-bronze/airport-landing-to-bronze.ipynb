 - read the landed airport reference file and writes it into bronze as a managed delta table
 - source: aeropulse_landing_lh -> Files/airport/
 - target: aeropulse_bronze_lh -> bronze_airport
 - load type: full - entire table overwritten on every run
 - dependencies: bronze-environment, bronze-helpers

## create airport_df schema

```python
airport_schema = StructType(
    [
     StructField('Code', StringType()), 
     StructField('Description', StringType()), 
     StructField('batch_id', StringType()), 
     StructField('ingested_timestamp', StringType()), 
     StructField('source_path', StringType())
    ]
)
```

## read landing lakehouse airport.csv

```python
airport_df = spark.read.format('csv').option("header", "true").schema(airport_schema).option("mode", 'FAILFAST').load(source_path)
```

## write landing airport_df to bronze layer

```python
write_to_bronze(
    airport_df,
    target_table = "bronze_airport",
    batch_id="batch_id",
    load_type ="full"
)
```
