## Landing to bronze: carriers
 - read the landed carrier reference file and writes it into bronze as a managed delta table
 - source: aeropulse_landing_lh -> Files/carrier/
 - target: aeropulse_bronze_lh -> bronze_carrier
 - load type: full - entire table overwritten on every run
 - dependencies: bronze-environment, bronze-helpers

```python
from pyspark.sql import functions as F
from pyspark.sql.types import *
```

## create an ingestion parameter

```python
batch_id = ""
batch_year = ""
```

## call environment variable

```python
%run bronze-environment
```

```python
source_path = f"{landing_lakehouse_path}/carrier/*.csv"
print(source_path)
```

## call write_to_bronze function

```python
%run bronze-helper
```

## create carrier_df schema

```python
carrier_schema = StructType([
    StructField('Code', StringType()), 
    StructField('Description', StringType()), 
    StructField('batch_id', StringType()), 
    StructField('ingested_timestamp', StringType()), 
    StructField('source_path', StringType())
    ]
)
```

## read landing carrier.csv file

```python
carrier_df = spark.read.format('csv').option("header", "true").schema(carrier_schema).option("mode", 'FAILFAST').load(source_path)
```

## writing carrier_df to bronze lakehouse

```python
write_to_bronze(
    carrier_df,
    target_table = "bronze_carrier",
    batch_id="batch_id",
    load_type ="full"
)
```
