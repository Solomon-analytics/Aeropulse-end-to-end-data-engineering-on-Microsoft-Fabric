## Aeropulse — Gold: Carrier Dimension

**Purpose:** Builds `dim_carrier` from silver `carrier`, dropping the lineage columns (`batch_id`, `ingested_timestamp`, `source_path`) that don't belong in a reporting-facing dimension.

**Batch parameters:** `batch_id`, `batch_year` — used to filter the silver `carrier` read

**Depends on:** `gold-environment`, `gold-helper` (run via `%run`)

**Reads:** `silver.carrier`

**Writes:** `dim_carrier` (merge on `carrier_sk`)


```python
from pyspark.sql import functions as F
from pyspark.sql.types import *
```

```python
batch_id = ""
batch_year = ""
```

```python
%run gold-environment
```

```python
%run gold-helper
```

## read carrier dataframe

```python
carrier_df = spark.read.format('delta').load(carrier_silver_path).filter(F.col("batch_id")==batch_id)
```

## drop unnecessary columns

```python
carrier_df = carrier_df.drop("batch_id", "ingested_timestamp", "source_path")
```

# write to gold layer

```python
# write to gold layer

update_cols = [c for c in carrier_df.columns if c not in ["carrier_sk"]]

write_to_gold(
    carrier_df,
    "dim_carrier",
    "s.carrier_sk = t.carrier_sk",
    update_cols
)
```
