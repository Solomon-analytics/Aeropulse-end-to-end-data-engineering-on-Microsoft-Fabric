## Aeropulse — Orchestration: 02 Create New Batch

**Purpose:** Marks a batch `IN_PROGRESS` and stamps its start time, immediately before the landing → bronze → silver → gold notebooks run for it.

**Parameters (pipeline-injected):** `source_name`, `batch_id`


```python
# Parameters
# Injected by the pipeline's Notebook activity base parameters — defaults below are for standalone testing only.
source_name = "flight"
batch_id = ""
```

```python
# 02 - create new batch (mark in progress)
# flip the batch to IN_PROGRESS and record when processing started —
# 03-complete-batch / 04-fail-batch close it out once the transform notebooks finish
spark.sql(f"""
    UPDATE control.batch_control
    SET status = 'IN_PROGRESS', start_time = current_timestamp(), updated_timestamp = current_timestamp()
    WHERE source_name = '{source_name}' AND batch_id = '{batch_id}'
""")
```
