## Aeropulse — Orchestration: 03 Complete Batch

**Purpose:** Marks a batch `COMPLETED` once landing → bronze → silver → gold have all finished successfully for it. Wire this to the *success* path of the last transform activity in the pipeline.

**Parameters (pipeline-injected):** `source_name`, `batch_id`


```python
# Parameters
# Injected by the pipeline's Notebook activity base parameters — defaults below are for standalone testing only.
source_name = "flight"
batch_id = ""
```

```python
# 03 - complete batch
# flip the batch to COMPLETED and record when it finished — only reached on the pipeline's success path
spark.sql(f"""
    UPDATE control.batch_control
    SET status = 'COMPLETED', end_time = current_timestamp(), updated_timestamp = current_timestamp()
    WHERE source_name = '{source_name}' AND batch_id = '{batch_id}'
""")
```
